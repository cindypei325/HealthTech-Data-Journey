今日實戰任務：自動化病患優先級標籤系統

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("healthcare_dataset.csv")
df
```

```python
#1. 計算全院的「平均醫療費用」

avg_billing = df['Billing Amount'].mean()
print(f"步驟一：全院平均醫療費用為: {avg_billing:.2f} 元")
```

```python
#2. 先幫所有人預設貼上 'Standard' (標準) 標籤
#利用Pandas的廣播機制 (Broadcasting)讓表格強制加上一欄，並填滿'Standard'
#預設值 (Default Value)」防禦性寫法：先假設所有人都是標準狀態，再把特殊的高風險病人挑出來修改

df['Health_Priority'] = 'Standard'
df.head()
```

```python
# 3. 設定多重條件！
# 關鍵細節：在 Pandas 裡，兩個條件之間要用 & 連結，而且每個條件都必須用小括號 () 包起來！
# Boolean Mask：並非把數據挑出來，而是做出一個核對表（Series 物件），每個都被填入True或False

high_risk_condition = (df['Billing Amount'] > avg_billing) & (df['Age'] > 60)

type(high_risk_condition)
high_risk_condition
```

```python
# 4. 瞄準並覆寫符合條件的人，改成 'High Attention' (高關注)
#把剛剛那份清單用loc的方式找到high_risk_condition=true（默認）的，把他們的Health_Priority都改成High Attention

df.loc[high_risk_condition, 'Health_Priority'] = 'High Attention'
df
```

```python
# 5. 驗收成果報告

print("\n📊 步驟二：標籤貼放完成！以下是全院病患優先級統計：")
print(df['Health_Priority'].value_counts())
```

```python
# 6. 工程師的防呆機制：隨機印出 5 個被標記為 'High Attention' 的病人，親眼核對邏輯對不對

print("\n👀 步驟三：抽查 5 位高關注病患 (確認年齡是否真的 > 60 且 費用 > 平均)：")
display(df[df['Health_Priority'] == 'High Attention'][['Age', 'Billing Amount', 'Health_Priority']].head())
```
