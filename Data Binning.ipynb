Feature Engineering -- Data Binning

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("healthcare_dataset.csv")
df
```

```python
# 1. 定義年齡的分段點與標籤
#像是拿了四個大箱子，第一個箱子裝0-18歲、第二個箱子裝18-35歲，以此類推（注意：五個數字、四個間隔！）
#之後分別在這些箱子上貼上標籤，第一個箱子貼上Pediatric、第二個箱子貼上Young Adult

bins = [0, 18, 35, 60, 100]
labels = ['Pediatric', 'Young Adult', 'Adult', 'Senior']
df.head()
```

```python
# 2. 建立新欄位 Age_Group
#把pd.cut視為一個切割機，機器讀取數據中的age欄位，看到數據後自動判斷在哪個箱子，並貼上標籤，最後加上Age_Group這個欄位

df['Age_Group'] = pd.cut(df['Age'], bins=bins, labels=labels)
df.head()
```

```python
# 3. 觀察每個族群的平均住院天數 (Length of Stay)
#之前算過的Recovery_Days：轉換日期資料、計算結果、去除days字眼

df['Date of Admission'] = pd.to_datetime(df['Date of Admission'])
df['Discharge Date'] = pd.to_datetime(df['Discharge Date'])
df['Recovery_Days'] = df['Discharge Date'] - df['Date of Admission']
df['Recovery_Days'] = (df['Discharge Date'] - df['Date of Admission']).dt.days
df.head()
```

```python
age_group_analysis = df.groupby('Age_Group', observed=True)['Recovery_Days'].mean()
print(age_group_analysis)

#出現問題：observed，確認是true，請列出真正有觀察到的！
```
